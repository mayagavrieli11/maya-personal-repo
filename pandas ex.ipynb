```python
import pandas as pd

# Create Customers DataFrame
customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106],
    "name": ["Noa", "Tom", "Dana", "Maya", "Yuval", "Daniel"],
    "age": [24, 31, 27, 25, 29, 35],
    "country": ["Israel", "Israel", "USA", "Israel", "UK", "USA"]
})

# Create Purchases DataFrame
purchases = pd.DataFrame({
    "purchase_id": [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    "customer_id": [101, 102, 101, 103, 105, 103, 106],
    "product": [
        "Laptop", "Mouse", "Keyboard", "Monitor",
        "Mouse", "Laptop", "Keyboard"
    ],
    "amount": [4200, 120, 350, 1500, 100, 3900, 300]
})

# Create Sessions DataFrame
sessions = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105, 106],
    "sessions": [12, 7, 18, 15, 5, 9],
    "minutes_on_site": [85, 32, 140, 110, 25, 67]
})

# Create Excel file with 3 sheets
file_name = "company_data.xlsx"

with pd.ExcelWriter(file_name, engine="openpyxl") as writer:
    customers.to_excel(
        writer,
        sheet_name="Customers",
        index=False
    )

    purchases.to_excel(
        writer,
        sheet_name="Purchases",
        index=False
    )

    sessions.to_excel(
    
        writer,
        sheet_name="Sessions",
        index=False
    )

# Success message
print(f"Excel file '{file_name}' was created successfully!")
```

General Data Analysis for excel file 

```python
import pandas as pd

file_path = "company_data.xlsx"

# Load ALL sheets
sheets = pd.read_excel(file_path, sheet_name=None)

print(f"Number of sheets: {len(sheets)}")
print(f"Sheet names: {list(sheets.keys())}")

# Run the same overview for EVERY sheet
for sheet_name, df in sheets.items():

    print("\n" + "=" * 60)
    print(f"SHEET: {sheet_name}")
    print("=" * 60)

    # Size
    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    # Column names
    print("\nColumn names:")
    print(df.columns.tolist())

    # Data types
    print("\nData types:")
    print(df.dtypes)

    # Missing values
    print("\nMissing values:")
    print(df.isna().sum())

    # Duplicates
    print(f"\nDuplicated rows: {df.duplicated().sum()}")

    # Number of unique values
    print("\nUnique values per column:")
    print(df.nunique())

    # First rows
    print("\nFirst 5 rows:")
    print(df.head())

    # Statistics
    print("\nStatistics:")
    print(df.describe(include="all"))
```
