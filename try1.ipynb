```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Ready for data analysis!")
```

```python
data = {
    "name": ["Noa", "Dana", "Tom", "Maya", "Yuval"],
    "age": [21, 31, 27, 25, 29],
    "sessions": [12, 7, 18, 15, 5],
    "purchases": [3, 1, 6, 4, 0]
}

df = pd.DataFrame(data)

df
```

```python
df.describe()
```

```python
df.groupby("age")["purchases"].mean()
```

```python
df.groupby("age")[["sessions", "purchases"]].mean()
```

```python
bins = [18, 25, 30, 35, 45, float("inf")]

labels = [
    "18-24",
    "25-29",
    "30-34",
    "35-44",
    "45+"
]
df["age_group"] = pd.cut(
    df["age"],
    bins=bins,
    labels=labels,
    right=False
)
df
```

```python
age_analysis = df.groupby("age_group", observed=True).agg(
    users=("name", "count"),
    total_sessions=("sessions", "sum"),
    avg_sessions=("sessions", "mean"),
    total_purchases=("purchases", "sum"),
    avg_purchases=("purchases", "mean")
)

age_analysis
```

```python
# מספר קבוצות הגיל שאנחנו רוצות
n_groups = 3

# יצירת קבוצות גיל עם מספר דומה של אנשים בכל קבוצה
age_bins = pd.qcut(
    df["age"],
    q=n_groups,
    duplicates="drop"
 )
df["age_group"] = age_bins.map(
    lambda age_range: f"{int(age_range.left)}-{int(age_range.right)}"
 )

# יצירת טבלת סיכום
age_analysis = (
    df.groupby("age_group", observed=True)
      .agg(
          users=("age", "count"),
          avg_age=("age", "mean"),
          total_sessions=("sessions", "sum"),
          avg_sessions=("sessions", "mean"),
          total_purchases=("purchases", "sum"),
          avg_purchases=("purchases", "mean")
      )
      .reset_index()
 )

age_analysis
```
